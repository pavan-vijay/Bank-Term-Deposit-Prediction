#### Outlier Analysis

```python
outlier_cols = dataset[['age', 'balance', 'duration',
                        'campaign', 'previous']]
```

```python
plt.figure(figsize=(15, 8))

for i, col in enumerate(outlier_cols.columns, 1):
    plt.subplot(2, 3, i)
    sns.boxplot(y=outlier_cols[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
for col in outlier_cols.columns:

    Q1 = outlier_cols[col].quantile(0.25)
    Q3 = outlier_cols[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outlier_count = ((outlier_cols[col] < lower) |
                     (outlier_cols[col] > upper)).sum()

    print("Column:", col)
    print("Lower limit:", lower)
    print("Upper limit:", upper)
    print("Outliers:", outlier_count)
    print("-----------------------")
```

```python
for col in outlier_cols.columns:
    print(col, "Minimum value:", dataset[col].min())
```

```python
print("Minimum pdays:", dataset['pdays'].min())
print("Maximum pdays:", dataset['pdays'].max())

print("Negative pdays:",
      (dataset['pdays'] < 0).sum())
```

```python
for col in outlier_cols.columns:
    print(col, "Skewness:", dataset[col].skew())
    
```

```python
for col in outlier_cols.columns:
    plt.figure(figsize=(6, 3))
    sns.histplot(dataset[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()



# The IQR method identified potential outliers in several numerical features. 
# However, these values may represent genuine customer information rather than data-entry errors. 
# Therefore, no observations were removed or capped during the initial EDA.
# The impact of outlier treatment will be evaluated during preprocessing and model comparison.
```
